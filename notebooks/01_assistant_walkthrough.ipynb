# AI-Powered IT Troubleshooting Assistant

This notebook demonstrates safe, structured troubleshooting guidance. It does not execute commands, request passwords, or claim a fixed accuracy benchmark. The deterministic path runs without an API key.

```python
from src.assistant import TroubleshootingAssistant

assistant = TroubleshootingAssistant()
examples = [
    'My Wi-Fi is connected but websites will not load',
    'I am locked out and cannot sign in',
    'My computer is very slow today',
]
for example in examples:
    print(example)
    print(assistant.respond(example))
```

## Response contract

Each response contains an intent, a human-readable message, ordered steps, escalation guidance, a confirmation flag, and an empty `executed_actions` list. The empty list is intentional: this project provides guidance and does not make system changes.

```python
reply = assistant.respond('The network is down on my laptop')
assert reply['intent'] == 'wifi'
assert reply['requires_confirmation'] is True
assert reply['executed_actions'] == []
print(reply)
```

## Optional LLM boundary

An optional OpenAI-compatible adapter is kept separate from the deterministic playbooks. Configure `LLM_BASE_URL`, `LLM_API_KEY`, and `LLM_MODEL` only through environment variables. The adapter parses JSON and fails closed when configuration or response shape is invalid. Never place secrets in a notebook, commit them, or send passwords and recovery codes to a model.
